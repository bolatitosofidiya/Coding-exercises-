# String Formatting 

#### Interpolation
#### This is the insertion of something of a different nature into something else

#### String Interpolation and String formatting are the same
#### String formatting is the process of inserting an object into a predefined string 

### %formatting
#### This is the oldest technique used to insert an object into a string
#### Multiple variable are allowed
#### %s is called a place holder

```python
name = "Angela"
print("My name is %s." % name)
```

```python
name = "Angela"
city ="Lagos"
print("My name is %s and l live in %s." % (name, city))
```

#### %f is a place holder for floating value
#### %d is a place holder for integar value
#### %b is a place holder for binary

```python
acc = 300
num = 400
print("My account balance is %d and l need %d" %(acc, num))
```

### str.format() function
#### The placeholder are replaced by curly braces in this technique.
#### We can reference variable through indexing

```python
# curly brace 
sport = "relay"
sex = "Male"

print("We are have {} for the {}.".format(sport, sex))
```

```python
# indexing

print("We are have {1} for the {0}.".format(sport, sex))
```

```python
n = "Jacob"
c ="Ibadan"
print("My name is {name} and l live {city}".format(name=n, city=c))
```

```python
print("I got {0:f}% marks in english.".format(55.66))
```

```python
# The length of the value after the decimal can be controlled

print("I got {0:.2f}% marks in english.".format(55.645686))
```

### F string
#### This is use to format string

```python
# Multiline f string

name = "Bree"
age = 24
occupation = "Doctor"

print(f"My name is {name}.\n"
     f"I am {age}.\n"
     f"I am a {occupation}.")
```

```python
print(f"My name is {name}.\n"
      f"I am {age} years.\n"
      f"I am a {occupation.upper()}")
```

```python
intro = f"""My name is {name}.
I am {age} years old.
I am a "{occupation}."""

print(intro)
```
