#### String Slicing
#### This is a technique used to access a portion (substring) of a string by specifying where to start and stop.
#### Substring is a smaller section taken from a large string.

#### Slcing in python has 2 part
#### Where to start
#### Where to stop (This stop before the last character)
#### There is a 3rd parameter:(step value)
#### The 3rd parameter allows us to control how python move through the string
####We can tell python to:
#### Read every second character 
#### Read every third character 
#### Read every fourth character
#### Skip character while reading

```python
# The first index tells python where to start
# The second index tells python where to stop -1
# The last number tells python to move to the every 2nd number 

job = "Python"
print(job[0:7:2])
```

```python
print(job[::2])
```

```python
country = "nigeria"
print(country[1::2])
```

```python
text = "pythonprogramming"

print(text[::3])
```

#### Negative Step Value
#### Python now start from the end and walk towards the begining

```python
word = "python"
print(word[::-1])
```

```python
# it start position is usually greater than the stop position
tile = "program"
print(tile[5:1:-1])
```

```python
word ="ANALYTICS"
print(word[7:2:-1])
```

```python
# Using string slicing to extra first and last name

customer_name = input("Enter Fullname: ")

space = customer_name.find(" ")

print("First Name:",customer_name[:space])
print("Last Name:",customer_name[- space:])
```

```python
# Using string slicing to extra first, middle and last name

customer_name = input("Enter Fullname: ")

first_space = customer_name.find(" ")
second_space = customer_name.find(" ",first_space + 1)

print("First Name:", customer_name[:first_space])
print("Middle Name:",customer_name[first_space + 1 : second_space])
print("Last Name:",customer_name[second_space + 1:])
```

```python
# If you don't know whether there's a middle name

# Using split() is much easier:

customer_name = input("Enter Fullname: ")
part = customer_name.split()

print("First Name:",part[0])

if len(part) == 3:
    print("Second Name:",part[1])
    print("Last Name:",part[2])
    
    
elif len(part) == 2:
    print("Last Name:",part[1])
```
