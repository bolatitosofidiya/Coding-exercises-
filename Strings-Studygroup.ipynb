# String
#### String is simply a sequence of characters used to represent text in python. 
#### It can contain letters, numbers, space, punctuation marks or special symbol.
#### Python treats the entire sequence as one piece of text
#### Python allow string to be enclosed in either single quote (') or double quote ("). Both are valid and behave the same way.

#### Using Quotes inside string
#### Rule 1
#### If your message contains double quote use single quote for the inner string and vice versa

```python
# Use different type of quote
# surround the string with one type of quotation mark and use the other inside the string

sen = 'Python is"The King"'
print(sen)
```

#### If you want to use double quote twice in a string then
#### use escape character (\) it tells python to treat the following quotation mark as part of text.

```python
name = "He said \"I'm Bassy\""
print(name)
```

```python
job = "He said \"I'm learning Python\""
print (job)
```

```python
movie = "The movie is called \"Inception\"."
print(movie)
```

```python
age = 'I\'m a girl'
print(age)
```

### Multiline String
#### This is a string that can span multiple line while preserving the line break and formatting eactly as written.
#### They are enclosed in triple double quote("""""""") or triple single quote ('''''')

```python
message ="""Welcome to San Academy
We are exicted to have you
Happy Learning"""

print(message)
```

```python
email = "Hello\nBella\nRegards"
print(email)
```

```python
week = """Today is Tuesday.
I saw 'Tomi' yesterday"""
print(week)
```

```python
no ="50"
print(no)
```
