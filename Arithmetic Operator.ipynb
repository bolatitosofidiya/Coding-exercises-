# Arithmetic Operators

### Operator
##### This ia a special symbol that tell python what to do e.g +, -, /
##### example 5 + 3 (The + symbol is tell python to add both numbers together)

### Operands
#### The value that operator act upon are called operands
#### e.g 5 + 3 (5 is the first operand, 3 is the second operand and + is the operator)

### Expression 
#### This is any valid combination of operand, value , variable, and operator that python can evaluate to produce a result
#### 5+3, salary + bonus, 4+5*2 , all this are example of expression

### Unary Operators
#### This works with one operand by indicating or changing the sign of a number
#### e.g +x, -x, if x=10 then first output 10 while second output is -10

### Binary Operators
#### This works on 2 operands, i.e performing calculation between 2 value
#### e.g 10+5 (The addition sign works on the two operand)
