```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv("pizza_sales.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.plot(x="pizza_category", y="total_price",title="Pizza-Sales",legend=True)
```

```python
df["T-U"] = df.total_price - df.unit_price
```

```python
df['100MA'] = df['order_id'].rolling(100).mean()
```

```python
sns.set_style('darkgrid')
```

```python
ax = plt.axes(projection='3d')
ax.scatter(df.index,df['T-U'],df['100MA'])
ax.set_xlabel('Index')
ax.set_ylabel('T-U')
ax.set_zlabel('100MA')
plt.show()
```

```python
def return_z(x,y):
    return 50 -(x**2+y**2)
```

```python
sns.set_style('whitegrid')
```

```python
x1,y1 =np.linspace(-5,5,50), np.linspace(-5,5,50)
```

```python
x1,y1 =np.meshgrid(x1,y1)
z1= return_z(x1,y1)
```

```python
ax =plt.axes(projection='3d')
ax.plot_surface(x1,y1,z1)
plt.show()
```
