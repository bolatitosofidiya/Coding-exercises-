#### There are 4 built-in data structure
##### 1. List
##### 2. Tuple
##### 3. Set
##### 4. Dictionary

##### List [ ] - ordered collection of item, can store duplicate, can be modify, maintain order.
##### Tuple ( )- ordered collection that cannot be changed, can store duplicate,cannot be modify, maintain order.
##### Set { }- unordered collection of unique items, can't store duplicate, can be modify, does not mantain order.
##### Dictionary { }- collection of key-value pair, keys can't store duplicate while value can store duplicate, can be modify, maintain order.


##### If you need to 
##### Store item that may change use LIST
##### Store item that should never change use TUPLE
##### Store unique value use SET
##### Store information using label use dictionary

#### Function
##### This is a reusable block of code that perform specific task. Function usually accept one or more input(argument), perform an operation
##### and return a result.
##### general syntax
##### function_name(argument)
##### Example:
##### score = [10,20,30,40]
##### print(len(score))
##### print(sum(score))
##### print(max(score))
##### print(min(score))
##### Output
##### 4
##### 100
##### 40
##### 10

#### Method
##### This is similar to a function, but it belong to a specific object or data structure.
##### Instead of  placing data inside a ( ) method are called directly on object using dot(.)
##### General Syntax:
##### object.method()
##### Example:
##### fruit = ["Apple", "Banana"]
##### fruit.append("Orange")
##### Here append() is a method
##### It belongs to list
##### It adds a new item directly to the existing list.
##### The append operation is perform using a method because it belong specifically to list object

##### Function                                                                  Method
##### Independent of any object                          Belongs to a specific object
##### Object is passed as an argument               Called directly on the object
##### Usually return a new result                        Often change the original object
##### Syntax: 
##### function(object)                                            Object.method()
