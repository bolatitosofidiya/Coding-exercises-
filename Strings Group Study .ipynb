## Strings
#### This is a sequence of character enclosed in quotation marks.
#### It may contain letters, numbers, symbols, space, emoji

## String Concatenation
#### This is the joining of two or more string together
#### python use "+" operator for concatenation.

```python
name = "John" + " " + "Smith"
print(name)
```

```python
first_name ="Jon"
last_name = "Williams"

fullname = first_name + " " + last_name
print(fullname)
```

#### String formating with ".format()"

```python
name = "Jane"
age = 34

print("Hello {}, you are {} years." .format(name, age))
```

#### Using .format improve readability when inserting many variable

#### Using f-string

```python
name = "Bolu"
job ="Doctor"

print(f"My name is {name}, l work as a {job}.")
```

#### F-strings is preferred because
#### They are faster, cleaner, easier to read, easier to maintain
#### Most python developer prefer f-strings over .format()

#### String Indexing
#### Every character inside a string has a position called an index
#### python start from 0
#### Example : word ="Hello"
#### H e l l o
#### 0 1 2 3 4

```python
word = "Hello"
print(word[0])
```

#### Negartive Indexing
#### This counts from the end
#### Example: 
####  H  e  l  l  o
#### -5 -4 -3 -2 -1

```python
print(word[-4])
```

#### String Slicing
#### This extract part of a string

#### Syntax: string[start: stop]
#### The stop position is not included

```python
# python includes the index 0 and 1 and ignore 2
print(word[0:2])
```

```python
# Entire word
print(word[:])
```

```python
# Begining to end
print(word[2:])
```

```python
# Begining to end
print(word[0:])
```

#### Using Steps
#### Syntax: String[start:stop:step]

```python
print(word[0:5:2])
```

#### Replacing Text with replace()
#### Syntax: String.replace(old, new)

```python
text ="Hello World!"

print(text.replace("!", "."))
```

```python
print(text.replace("World", "Nigeria"))
```

#### Finding text with find()
#### This return the position where a word or character first appears.
#### Synatax: string.find(value)

```python
sentence = "Look around what is going on"
print(sentence.find("what"))
```

```python
# Character Search
text ="python"

print(text.find("t"))
```

```python
# When an item does not exist python print -1

print(text.find("b"))
```

```python
# Removing extra space using strip()

name ="  Joshua  "

print(name)
```

```python
print(name.strip())
```

#### Other variant
#### Remove left space with name.lstrip()

#### Remove right space with name.rstring()

```python
# Example
name = "    Loveth    "

print(name.lstrip())
```

```python
print(name.rstrip())
```

#### Splitting Strings with split()
#### This method divide a string into multiple parts.
#### Syntax: String.split()

```python
# python create a list
text = "Today is Saturday"

print(text.split())
```

```python
# Splitting by comma
name ="John,Mary, Peter,Joshua"

print(name.split(","))
```

```python
# In excel proper is used to capitalize each word but python used title
name = "my name is Great"

print(name.title())
```
