### Operator Precedence
#### Python evaluate logical operators in the following order:
##### 1. Parentheses ()
##### 2. not means reversal e.g True changes to False and vice versa
##### 3. and
##### 4. or


##### so this:
###### x==5 or y > 5 and z < 4

##### is interpreted as:
###### x ==5 or(y > 5 and z < 4)
##### because "and" has higher precedence than or

```python
print(x==5 or y>5 and z<4)
```

### Why Parentheses matters

## Best practice

#### Use "in" when checking whether something exist in a collection
#### Use "not in" when checking whether something is absent 
#### Use "==" for value comparison
#### Use "is" only for identity comparison
#### Use "is not" when comparing different identity or confirming that a value exist... e.g if result is not None
