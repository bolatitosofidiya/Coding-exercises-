```python
df['V-Q'] =df.Value -df.Quantity
df['100MA'] =df['Quantity'].rolling(100).mean()
```

```python
ax =plt.axes(projection='3d')
ax.scatter(df.index,df['V-Q'], df['100MA'])
ax.set_xlabel('Index')
ax.set_ylabel('V-Q')
ax.set_ylabel('100MA')
plt.show()
```
